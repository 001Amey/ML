## 1. Univariant Analysis


```python
import matplotlib.pyplot as plt
import seaborn as sns

numeric_cols = data.columns

for col in numeric_cols:
    plt.figure(figsize=(7,4))
    sns.histplot(data[col],kde=True)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()
```

```python
# Boxplot _OUTLIER DETECTION
for col in numeric_cols:
    plt.figure(figsize=(7,3))
    sns.boxplot(x=data[col])
    plt.title(f'Boxplot of{col}')
    plt.show()
```

```python
features= data.drop(columns=['quality'])

Q1=features.quantile(0.25)
Q3=features.quantile(0.75)

IQR= Q3+Q1
lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR

outlier_mask=(features<lower_bound)| (features>upper_bound)

outlier_count = outlier_mask.sum()

outlier_percentage = (outlier_count / len(features))* 100
outlier_summary = pd.DataFrame({
    'Q1': Q1,
    'Q3': Q3,
    'IQR':IQR,
    'Lower Bound':lower_bound,
    'Upper Bound': upper_bound,
    'Outlier Count': outlier_count,
    'Outlier %' : outlier_percentage,
    
})
outlier_summary.round(3)
```

```python
plt.figure(figsize=(12,6))

outlier_count.sort_values(ascending=False).plot(
    kind='bar'
)
plt.title("Number of outlier in each Feature")

plt.xlabel("Features")
plt.ylabel("Number of Outliers")
plt.xticks(rotation=45,ha='right')
plt.grid(axis='y',alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier , plot_tree
from sklearn.metrics import (accuracy_score,precision_score,recall_score,f1_score,classification_report,confusion_matrix,ConfusionMatrixDisplay)

```

```python
X=data.drop('quality',axis=1)
y=data['quality']
print('X shape:',X.shape)
print('y shape:',y.shape)
```

```python
print(y.value_counts().sort_index())
```

```python
X_train,X_test,y_train,y_test=train_test_split(X , y , test_size=0.20, random_state=42, stratify=y)
```

```python
criterion='entropy'
dt_model=DecisionTreeClassifier(criterion='entropy',random_state=42)
```

```python
data['quality_group']=pd.cut(
    data['quality'],
    bins=[0,5,6,10],
    labels=['low','medium','high']
)
```

```python
print(X.columns)
```

```python
print(X.select_dtypes(exclude='number').columns)
```

```python
y=data['quality']
```

```python
X=data.drop(columns=['quality','quality_group'],errors ='ignore')
```

```python
print(X.columns)
print(X.dtypes)
```

```python
X_train,X_test,y_train,y_test=train_test_split(X , y , test_size=0.20, random_state=42, stratify=y)
```

```python
print("X_train:",X_train.shape)
print("X_test:",X_test.shape)
print("y_train:",y_train.shape)
print("y_test:",y_test.shape)
```

```python
dt_model.fit(X_train,y_train)
```

```python
y_train_pred=dt_model.predict(X_train)
y_test_pred=dt_model.predict(X_test)
```

```python
train_accuracy=accuracy_score(y_train,y_train_pred)
print("Training Accuracy",train_accuracy)
```

```python
test_accuracy=accuracy_score(y_test,y_test_pred)
print("Testing Accuracy",test_accuracy)
```

```python
print("Precision :",precision_score(y_test,y_test_pred,average='weighted'))
print("Recall:",recall_score(y_test,y_test_pred,average='weighted'))
print("F1 score:",f1_score(y_test,y_test_pred,average='weighted'))
```

```python
print(classification_report(y_test,y_test_pred))
```

```python
# Confusion MAtrix
cm=confusion_matrix(y_test,y_test_pred)
plt.figure(figsize=(8,6))
sns.heatmap(cm,annot=True,fmt='d',cmap='Blues')

plt.xlabel('Predicted Quality')
plt.ylabel('Actual Quality')
plt.title("Confusion Matrix - TDecision Tree")
plt.show()
```

```python
print("Tree Depth:",dt_model.get_depth())
print("Number of leaves :",dt_model.get_n_leaves())
```

```python
plt.figure(figsize=(25,15))
plot_tree(
    dt_model,
    feature_names = X.columns,
    class_names=[str(c)for c in sorted(y.unique())],
    filled=True,
    rounded=True,
    fontsize=8
)
plt.title("Decision Tree - Entropy")
plt.show()
```

```python
plt.figure(figsize=(24,12))
plot_tree( dt_model,
    feature_names = X.columns,
    class_names=[str(c)for c in sorted(y.unique())],
    filled=True,
    rounded=True,
    fontsize=8,
    impurity=True,
    precision = 2,
    max_depth=3
)
plt.title("Decision Tree -Top- 3 Levels",fontsize=18)
plt.tight_layout()
plt.show()
```
